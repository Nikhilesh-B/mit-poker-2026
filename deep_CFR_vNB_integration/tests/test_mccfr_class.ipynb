# Test MCCFR Class

This notebook tests the MCCFR class to verify it works correctly.

```python
import sys
import os

# In Jupyter notebooks, __file__ is not defined
# Add parent directory to path so we can import from custom_engine
current_dir = os.getcwd()
if current_dir.endswith('deep_CFR_vNB_integration'):
    # We're in the integration directory, go up one level
    parent_dir = os.path.dirname(current_dir)
else:
    # Try to find the parent directory
    # Go up until we find mit-poker-2026
    parent_dir = current_dir
    while not parent_dir.endswith('mit-poker-2026') and parent_dir != os.path.dirname(parent_dir):
        parent_dir = os.path.dirname(parent_dir)
    if not parent_dir.endswith('mit-poker-2026'):
        # Fallback: assume we're in integration, go up one level
        parent_dir = os.path.dirname(current_dir)

sys.path.insert(0, parent_dir)

from mccfr import MCCFR
```

## Basic Functionality Tests

```python
print("=" * 70)
print("Testing MCCFR Class")
print("=" * 70)

# Initialize MCCFR
mccfr = MCCFR()
print("✓ MCCFR initialized")
```

```python
# Test creating initial state
state = mccfr.create_initial_state()
print(f"✓ Created initial state: street={state.street}, button={state.button}")
print(f"  Hand 0: {[str(c) for c in state.hands[0]]}")
print(f"  Hand 1: {[str(c) for c in state.hands[1]]}")
```

```python
# Test getting infoset
infoset = mccfr.get_infoset(state, player=0)
print(f"✓ Got infoset: {infoset[:50]}...")
```

```python
# Test getting legal actions
legal_actions = mccfr.get_legal_actions_list(state)
print(f"✓ Got {len(legal_actions)} legal actions")
for i, action in enumerate(legal_actions[:5]):  # Show first 5
    action_key = mccfr.action_to_key(action, state, 0)
    print(f"  {i+1}. {action_key}")
```

```python
# Test regret matching
test_regrets = {'FOLD': -10.0, 'CALL': 5.0, 'RAISE_SMALL': 15.0}
test_actions = [action for action in legal_actions if mccfr.action_to_key(
    action, state, 0) in test_regrets]
strategy = mccfr.regret_matching(test_regrets, test_actions, state, 0)
print(f"✓ Regret matching works:")
for key, prob in strategy.items():
    print(f"  {key}: {prob:.2%}")
```

## Single MCCFR Iteration Test

```python
print("\n" + "=" * 70)
print("Testing Single MCCFR Iteration")
print("=" * 70)

initial_state = mccfr.create_initial_state()
utility_p0 = mccfr.external_sampling(initial_state, traversing_player=0)
utility_p1 = mccfr.external_sampling(initial_state, traversing_player=1)

print(f"✓ Single iteration complete")
print(f"  Utility P0: {utility_p0:.2f}")
print(f"  Utility P1: {utility_p1:.2f}")
print(f"  Infosets visited: {len(mccfr.regret_table)}")
```

## Action Selection Test

```python
print("\n" + "=" * 70)
print("Testing Action Selection")
print("=" * 70)

action = mccfr.select_action(state, player=0, use_average=False)
action_key = mccfr.action_to_key(action, state, 0)
print(f"✓ Selected action: {action_key}")

print("\n" + "=" * 70)
print("✓ All Tests Passed!")
print("=" * 70)
```
